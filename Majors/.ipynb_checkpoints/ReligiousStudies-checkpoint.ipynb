<div class="text-box">

Here you'll find the breakdown of earned doctorates in Middle Eastern Studies 
    
The graphs below display demographic trends for the for the following CIP areas of study from the IPEDS survey of earned doctorates. 
    
**1) 16.11 - Middle/Near Eastern and Semitic Languages, Literatures, and Linguistics.**
    
**2) 05.0108 - Near and Middle Eastern Studies.**
    

</div>

```python
#| echo: false
import pandas as pd
import altair as alt

df=pd.read_csv('filtered_data.csv')
```

<div class="text-box">
    
## Earned Doctorates in Middle Eastern Studies by Race/Ethnicity 

```python
#| code-fold: true


majors = [
   '16.11 - Middle/Near Eastern and Semitic Languages, Literatures, and Linguistics.', 
'05.0108 - Near and Middle Eastern Studies.'

]

majors_df = (
     df[df["Major"].isin(majors)]
    .groupby(["Major", "Race and Ethnicity"])["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)


overall = (
    majors_df.groupby("Race and Ethnicity")["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)
overall["Major"] = "All Earned Doctorates in Middle Eastern Studies (total)"
majors_df = pd.concat([majors_df, overall], ignore_index=True)


major_dropdown = alt.binding_select(
    options=majors_df["Major"].unique().tolist(),
    name="Select Major: "
)
major_param = alt.param(
    name="Major",
    bind=major_dropdown,
    value="All Earned Doctorates in Middle Eastern Studies (total)"
)


chart = (
    alt.Chart(majors_df)
    .add_params(major_param)
    .transform_filter(
        alt.datum.Major == major_param
    )
    .mark_bar()
    .encode(
        x=alt.X("Awards/Degrees Conferred:Q", title="Degrees Conferred"),
        y=alt.Y("Race and Ethnicity:N", sort="-x", title="Race / Ethnicity"),
        color=alt.Color("Race and Ethnicity:N", legend=None),
        tooltip=["Race and Ethnicity:N", "Awards/Degrees Conferred:Q"]
    )
    .properties(
        title="Earned Doctorates in Middle Eastern Studies(2021–2023)",
        width=600,
        height=300
    )
)



chart
```

</div>

<div class="text-box">
    
## Earned Doctorates in Middle Eastern Studies by Sex 

```python
#| code-fold: true


gender_df = df[df["Major"].isin(majors)]


gender_agg = (
    gender_df.groupby("Sex")["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)


pie_chart = alt.Chart(gender_agg).mark_arc().encode(
    theta=alt.Theta(field="Awards/Degrees Conferred", type="quantitative"),
    color=alt.Color(field="Sex", type="nominal", legend=alt.Legend(title="Gender")),
    tooltip=["Sex", "Awards/Degrees Conferred"]
).properties(
    title="Gender Breakdown of Earned Doctorates in Middle Eastern Studies (2021–2023)",
    width=400,
    height=400
)

pie_chart
```

</div>

<div class="text-box">
    
## Top Institutions Awarding Doctorates in Middle Eastern Studies per Race/Ethnicity 

```python
#| code-fold: true


filtered_df = df[
    (df["Major"].isin(majors))]

grouped = (
    filtered_df.groupby(["Race and Ethnicity", "Institution Name"])["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)

grouped["Rank"] = grouped.groupby("Race and Ethnicity")["Awards/Degrees Conferred"]\
                         .rank(method="first", ascending=False)


top10 = grouped[grouped["Rank"] <= 10]


race_dropdown = alt.binding_select(
    options=top10["Race and Ethnicity"].unique().tolist(),
    name="Select Race/Ethnicity: "
)
race_param = alt.param(bind=race_dropdown, name="RaceFilter", value="Black or African American")


chart = (
    alt.Chart(top10)
    .add_params(race_param)
    .transform_filter(alt.datum["Race and Ethnicity"] == race_param)
    .mark_bar()
    .encode(
        x=alt.X("Awards/Degrees Conferred:Q", title="Degrees Conferred"),
        y=alt.Y("Institution Name:N", sort='-x', title="Institution"),
        color=alt.Color("Institution Name:N", legend=None),
        tooltip=["Institution Name", "Awards/Degrees Conferred"]
    )
    .properties(
        title="Top Institutions Awarding Doctorates in Middle Eastern Studies (2021-2023)",
        width=600,
        height=300
    )
)

chart
```
