The graphs below display demographic trends for the for the following CIP area of study from the IPEDS survey of earned doctorates.

```python
#| echo: false
import pandas as pd
import altair as alt
df=pd.read_csv('filtered_data.csv')
```

<div class="text-box">
    
## Earned Doctorates in Philosophy by Race/Ethnicity

```python
#| code-fold: true

majors = ['38.01 - Philosophy.']


filtered_df = df[
    (df["Major"].isin(majors))
]

grouped_df = (
    filtered_df.groupby("Race and Ethnicity")["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)


chart = (
    alt.Chart(grouped_df)
    .mark_bar()
    .encode(
        x=alt.X("Awards/Degrees Conferred:Q", title="Degrees Conferred"),
        y=alt.Y("Race and Ethnicity:N", sort="-x", title="Race / Ethnicity"),
        color=alt.Color("Race and Ethnicity:N", legend=None),
        tooltip=["Race and Ethnicity:N", "Awards/Degrees Conferred:Q"]
    )
    .properties(
        title="Earned Doctorates in Philosophy by Race/Ethnicity (2021–2023)",
        width=600,
        height=300
    )
)

chart
```

</div>

<div class="text-box">
    
## Earned Doctorates in Philosophy by Sex 
    
    
(Add a note here that IPEDS only includes Sex, and not a diverse array of gender identities)

```python
#|code-fold: true



gender_df = df[df["Major"].isin(majors)]


gender_agg = (
    gender_df.groupby("Sex")["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)


pie_chart = alt.Chart(gender_agg).mark_arc().encode(
    theta=alt.Theta(field="Awards/Degrees Conferred", type="quantitative"),
    color=alt.Color(field="Sex", type="nominal", legend=alt.Legend(title="Gender")),
    tooltip=["Sex", "Awards/Degrees Conferred"]
).properties(
    title="Gender Breakdown of Earned Doctorates in Philosophy (2021–2023)",
    width=400,
    height=400
)

pie_chart
```

</div>

## Top Instiutions Awarding Doctorates in Philosophy per Race/Ethnicity 

```python
#| code-fold: true

grouped = (
    filtered_df.groupby(["Race and Ethnicity", "Institution Name"])["Awards/Degrees Conferred"]
    .sum()
    .reset_index()
)

grouped["Rank"] = grouped.groupby("Race and Ethnicity")["Awards/Degrees Conferred"]\
                         .rank(method="first", ascending=False)


top10 = grouped[grouped["Rank"] <= 10]


race_dropdown = alt.binding_select(
    options=top10["Race and Ethnicity"].unique().tolist(),
    name="Select Race/Ethnicity: "
)
race_param = alt.param(bind=race_dropdown, name="RaceFilter", value="Black or African American")


chart = (
    alt.Chart(top10)
    .add_params(race_param)
    .transform_filter(alt.datum["Race and Ethnicity"] == race_param)
    .mark_bar()
    .encode(
        x=alt.X("Awards/Degrees Conferred:Q", title="Degrees Conferred"),
        y=alt.Y("Institution Name:N", sort='-x', title="Institution"),
        color=alt.Color("Institution Name:N", legend=None),
        tooltip=["Institution Name", "Awards/Degrees Conferred"]
    )
    .properties(
        title="Top Institutions Awarding Doctorates in Philosophy (2021–2023) ",
        width=600,
        height=300
    )
)

chart
```
